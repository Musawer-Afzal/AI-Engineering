````python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
````

````python
var = np.linspace(1, 10, 20).reshape(4, 5)
var
````

````python
sns.heatmap(var)
````

Using dataset data

````python
data = sns.load_dataset("anagrams")
data.head()
````

````python
x = data.drop(columns = ["attnr"], axis = 1).head(10)
x.head()
````

````python
sns.heatmap(x)
````

vmin and vmax parameter, assign the minimum and maximum values

````python
sns.heatmap(var, vmin = 0, vmax = 20)
````

cmap parameter, change the color theme that is displayed

````python
sns.heatmap(var, cmap = "PuOr")
````

annot parameter, The numerical values will now be displayed in the heatmap

````python
sns.heatmap(var, cmap = "PuOr", annot = True)
````

We can also assign our own values to the heatmap

````python
var1 = np.linspace(1, 10, 10).reshape(2, 5)
````

````python
ar = np.array([["a1","a2","a3","a4","a5",], 
                ["b1","b2","b3","b4","b5"]])
ar
````

fmt parameter, In order for our own data to work we need to use the fmt which specifies the format the new data is coming in

````python
sns.heatmap(
    var1, 
    vmin = 0, 
    vmax = 10, 
    cmap = "PuOr", 
    annot = ar,
    fmt = "s"   # string
    )
````

annot_kws parameter, change the style of how the annotated values are displayed in heatmap

````python
y = {"fontsize": 10, "color": "red", "weight": "bold"}

sns.heatmap(
    var1, 
    vmin = 0, 
    vmax = 10, 
    cmap = "PuOr", 
    annot = True,
    annot_kws = y
    )
````

linewidth parameter, change the width of lines between values in heatmap

````python
y = {"fontsize": 10, "color": "red", "weight": "bold"}

sns.heatmap(
    var1, 
    vmin = 0, 
    vmax = 10, 
    cmap = "PuOr", 
    annot = True,
    annot_kws = y,
    linewidths = 3,
    )
````

linecolor parameter, change the color of linewidth

````python
y = {"fontsize": 10, "color": "red", "weight": "bold"}

sns.heatmap(
    var1, 
    vmin = 0, 
    vmax = 10, 
    cmap = "PuOr", 
    annot = True,
    annot_kws = y,
    linewidths = 1,
    linecolor = "magenta"
    )
````

cbar parameter, the color bar we see at the right side, we can make it dissapear and appear

````python
y = {"fontsize": 10, "color": "red", "weight": "bold"}

sns.heatmap(
    var1, 
    vmin = 0, 
    vmax = 10, 
    cmap = "PuOr", 
    annot = True,
    annot_kws = y,
    cbar = False
    )
````
