Pair Plot only works with numerical values

````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
var = sns.load_dataset("tips")
var
````

````python
sns.pairplot(
    data = var,
)

plt.show()
````

hue parameter, compare the values according to a middle value

````python
sns.pairplot(
    data = var,
    hue = "sex",
)

plt.show()
````

We can plot the graph with select values only

````python
sns.pairplot(
    data = var,
    vars = ["total_bill", "tip"],
    hue = "sex"
)

plt.show()
````

order parameter, Sort the display priority

````python
sns.pairplot(
    data = var,
    vars = ["total_bill", "tip"],
    hue = "sex",
    hue_order = ["Female", "Male"],
)

plt.show()
````

palette parameter, change the color theme

````python
sns.pairplot(
    data = var,
    vars = ["total_bill", "tip"],
    hue = "sex",
    hue_order = ["Female", "Male"],
    palette = {"Female": "skyblue", "Male": "seagreen"}
)

plt.show()
````

x_vars and y_vars, we can decide which values should be used for plotting on x and y axis

````python
sns.pairplot(
    data = var,
    hue = "sex",
    hue_order = ["Female", "Male"],
    x_vars = ["total_bill", "tip", "size"],
    y_vars = ["total_bill", "tip"],
)

plt.show()
````

kind paramter, we can dictate which plot should be used to plot the values

We can use, "reg", "scatter", "kde", "hist"

````python
sns.pairplot(
    data = var,
    hue = "sex",
    hue_order = ["Female", "Male"],
    kind = "kde"
)

plt.show()
````

diag_kind parameter, we can tell which plot to use in diagonal

````python
sns.pairplot(
    data = var,
    hue = "sex",
    hue_order = ["Female", "Male"],
    diag_kind = "kde"
)

plt.show()
````

markers parameter, change the style of markers of labels

````python
sns.pairplot(
    data = var,
    hue = "sex",
    hue_order = ["Female", "Male"],
    markers = ["o", "*"],
)

plt.show()
````
