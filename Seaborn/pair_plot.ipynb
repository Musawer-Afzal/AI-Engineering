Pair Plot only works with numerical values

````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
var = sns.load_dataset("tips")
var
````

````python
sns.pairplot(
    data = var,
)

plt.show()
````

hue parameter, compare the values according to a middle value

````python
sns.pairplot(
    data = var,
    hue = "sex",
)

plt.show()
````

We can plot the graph with select values only

````python
sns.pairplot(
    data = var,
    vars = ["total_bill", "tip"],
    hue = "sex"
)

plt.show()
````

order parameter, Sort the display priority

````python
sns.pairplot(
    data = var,
    vars = ["total_bill", "tip"],
    hue = "sex",
    hue_order = ["Female", "Male"],
)

plt.show()
````

palette parameter, change the color theme

````python
sns.pairplot(
    data = var,
    vars = ["total_bill", "tip"],
    hue = "sex",
    hue_order = ["Female", "Male"],
    palette = {"Female": "skyblue", "Male": "seagreen"}
)

plt.show()
````
