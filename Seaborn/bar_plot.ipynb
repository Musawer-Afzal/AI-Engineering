````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
var = sns.load_dataset("penguins")
var.head()
````

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
)

plt.show()
````

hue parameter, divide the data using a parameter

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex"
)

plt.show()
````

order parameter, change the order of the data being displayed

````python
order_1 = ["Dream", "Torgersen", "Biscoe"]
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex",
    order = order_1
)

plt.show()
````

hue_order, change how the bars are being displayed according to the value given to hue

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex",
    hue_order = ["Female", "Male"],
)

plt.show()
````

ci parameter, Change the size of line at the top of every bar

use
```
errorbar = ("ci", int_value)
```

The range goes from 0 to 100

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex",
    errorbar = ("ci", 10)
)

plt.show()
````

n_boot parameter, change the values on y_axis

````python
sns.barplot(
    x = "bill_length_mm",
    y = "island",
    data = var,
    hue = "sex",
    orient = "h",
    n_boot = 1000
)

plt.show()
````
