````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
var = sns.load_dataset("penguins")
var.head()
````

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
)

plt.show()
````

hue parameter, divide the data using a parameter

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex"
)

plt.show()
````
