# Bar Plot

```
df = sns.load_dataset("")
sns.barplot(x = "", y = "", data = df)
plt.show()
```

````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
var = sns.load_dataset("penguins")
var.head()
````

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
)

plt.show()
````

hue parameter, divide the data using a parameter

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex"
)

plt.show()
````

order parameter, change the order of the data being displayed

````python
order_1 = ["Dream", "Torgersen", "Biscoe"]
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex",
    order = order_1
)

plt.show()
````

hue_order, change how the bars are being displayed according to the value given to hue

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex",
    hue_order = ["Female", "Male"],
)

plt.show()
````

ci parameter, Change the size of line at the top of every bar

use
```
errorbar = ("ci", int_value)
```

The range goes from 0 to 100

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex",
    errorbar = ("ci", 10)
)

plt.show()
````

n_boot parameter, change the values on y_axis

````python
sns.barplot(
    x = "bill_length_mm",
    y = "island",
    data = var,
    hue = "sex",
    orient = "h",
    n_boot = 1000
)

plt.show()
````

color parameter, change the color of bars

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex",
    color = "green"
)

plt.show()
````

palette parameter, use a compeltely differnt theme of color

Check other themes from the official website

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex",
    palette = "icefire"
)

plt.show()
````

saturation paramter, make the bars colors more vibrant or dull. It goes from 0 to 1

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex",
    saturation = 0.2
)

plt.show()
````

errcolor parameter, change the color of black lines on top of every bar

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex",
    errcolor = "magenta"
)

plt.show()
````

errwidth parameter, change the width of the line at the top of every bar

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex",
    errwidth = 5
)

plt.show()
````

capsize parameter, add a cap on the error line on top of every bar. Goes from 0 to 10

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex",
    capsize = 0.5
)

plt.show()
````

dodge parameter, superimpose more than one bars on top of eachother

````python
sns.barplot(
    x = "island",
    y = "bill_length_mm",
    data = var,
    hue = "sex",
    dodge = False,
    palette = "pastel"
)

plt.show()
````
