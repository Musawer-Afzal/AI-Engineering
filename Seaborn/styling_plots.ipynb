# Styling plots

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
var = sns.load_dataset("tips")
var
```

```python
sns.boxplot(
    y = var["total_bill"],
)

plt.show()
```

**sns.set_style** function to apply different styles to a the plot

We can use, "white", "dark", "darkgrid", "whitegrid", "ticks"

```python
sns.set_style("darkgrid")

sns.barplot(
    x = "day",
    y = "total_bill",
    data = var,
)

plt.show()
```

**sns.despine** function, we can remove the top and right line of the plot itself

```python
sns.set_style("white")

sns.barplot(
    x = "day",
    y = "total_bill",
    data = var,
)
sns.despine()

plt.show()
```

**plt.figure** function, change the height and width of the graph

```python
sns.set_style("white")
plt.figure(figsize=(8, 6))

sns.barplot(
    x = "day",
    y = "total_bill",
    data = var,
)
sns.despine()

plt.show()
```

**sns.set_context** function, change the labels properties

We can use, "poster", "paper" etc

```python
sns.set_style("white")
sns.set_context("poster", font_scale=1)

sns.barplot(
    x = "day",
    y = "total_bill",
    data = var,
)

plt.show()
```

palette parameter, change the color theme

```python
sns.set_style("white")
sns.set_context("paper", font_scale=1)

sns.barplot(
    x = "day",
    y = "total_bill",
    data = var,
    hue = "sex",
    palette = "pastel"
)

plt.show()
```
