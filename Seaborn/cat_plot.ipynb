# Cat Plot

Factor Plot has been renamed to Cat Plot in latest versions of Seaborn

```
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("")
sns.catplot(x = "", y = "", data = df)
plt.show()
```

### What kind of plots can we make using Catplot in Seaborn

**Categorical scatterplots with catplot**
- stripplot() - wiht kind = "strip"
- swarmplot() - with kind = "swarm"

**Categorical distribution plots with catplot**
- boxplot() - with kind = "box"
- violinplot() - with kind = "violin"
- boxenplot() - with kind = "boxen"

**Categorical estimate plots with catplot**
- pointplot() - with kind = "point"
- barplot() - with kind = "bar"
- countplot() - with kind = "count"

````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
var = sns.load_dataset("tips")
var
````

````python
sns.catplot(
    x = "size",
    y = "tip",
    data = var,
    hue = "sex",
    kind = "box"
)

plt.show()
````
