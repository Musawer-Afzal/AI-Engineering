````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
var = sns.load_dataset("penguins")
var.head()
````

hue parameter, which category is differentiting the data

````python
sns.scatterplot(
    x = "bill_length_mm",
    y = "bill_depth_mm",
    data = var,
    hue = "sex"
)

plt.show()
````

style parameter, change the way dots are appearing

````python
sns.scatterplot(
    x = "bill_length_mm",
    y = "bill_depth_mm",
    data = var,
    hue = "sex",
    style = "sex",
)

plt.show()
````

size parameter, change the size of dots according to a category

sizes parameter, give a value of the size you need

````python
sns.scatterplot(
    x = "bill_length_mm",
    y = "bill_depth_mm",
    data = var,
    hue = "sex",
    size = "sex",
    sizes = (10, 50)
)

plt.show()
````

palette parameter, change the theme of color

````python
sns.scatterplot(
    x = "bill_length_mm",
    y = "bill_depth_mm",
    data = var,
    hue = "sex",
    palette = "gist_ncar",
)

plt.show()
````

alpha parameter, change the transparency of dots

````python
sns.scatterplot(
    x = "bill_length_mm",
    y = "bill_depth_mm",
    data = var,
    hue = "sex",
    alpha = 0.3
)

plt.show()
````
