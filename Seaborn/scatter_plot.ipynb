````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
var = sns.load_dataset("penguins")
var.head()
````

````python
sns.scatterplot(
    x = "bill_length_mm",
    y = "bill_depth_mm",
    data = var,
    color = "green"
)

plt.show()
````
