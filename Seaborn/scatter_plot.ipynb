````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
var = sns.load_dataset("penguins")
var.head()
````

hue parameter, which category is differentiting the data

````python
sns.scatterplot(
    x = "bill_length_mm",
    y = "bill_depth_mm",
    data = var,
    hue = "sex"
)

plt.show()
````

style parameter, change the way dots are appearing

````python
sns.scatterplot(
    x = "bill_length_mm",
    y = "bill_depth_mm",
    data = var,
    hue = "sex",
    style = "sex",
)

plt.show()
````
