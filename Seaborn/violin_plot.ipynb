# Violin Plot

A violin plot plays a similar role as a box and whisker plot. It shows the distribution of quantitative data across several levels of one(or more) categorical variables such that those distributions can be compared

```
import seaborn as sns
sns.set(style = "whilegrid")
f = sns.load_dataset("fmri")
sns.violinplot(x = "", y = "", data = f)
```

````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
var = sns.load_dataset("tips")
var
````

````python
sns.violinplot(
    x = "day",
    y = "total_bill",
    data = var
)
````

hue parameter, compare the values according to a middle value

````python
sns.violinplot(
    x = "day",
    y = "total_bill",
    data = var,
    hue = "time"
)
````

````python
sns.violinplot(
    x = "day",
    y = "total_bill",
    data = var,
    linewidth = 3
)

plt.show()
````

palette parameter, change the color theme of violin graphs

````python
sns.violinplot(
    x = "day",
    y = "total_bill",
    data = var,
    palette = "Set2"
)

plt.show()
````

order parameter, we can sort the data according to any other category

````python
sns.violinplot(
    x = "time",
    y = "total_bill",
    data = var,
    order = ["Dinner", "Lunch"]
)


plt.show()
````

saturation parameter, darken or brighten the colors

````python
sns.violinplot(
    x = "day",
    y = "total_bill",
    data = var,
    saturation = 0.3
)

plt.show()
````

color parameter, change the color

````python
sns.violinplot(
    x = "day",
    y = "total_bill",
    data = var,
    color = "red"
)

plt.show()
````

split parameter, we can actually see the split of categories in the same violin plot

Use it with **hue** for better visualization

````python
sns.violinplot(
    x = "day",
    y = "total_bill",
    data = var,
    hue = "sex",
    split = True,
)

plt.show()
````

scale parameter, we can make changes to the violins for better visualization. We can use "count", "width", "area"

- "Count" will show the frequency of a particular category. How many were there.

````python
sns.violinplot(
    x = "day",
    y = "total_bill",
    data = var,
    hue = "sex",
    split = True,
    scale = "count",
)

plt.show()
````

In order to make the plot horizontal, just swap the values in x and y

````python
sns.violinplot(
    x = "total_bill",
    y = "day",
    data = var,
    hue = "sex",
)

plt.show()
````

inner paramter, change the inner box plot.

We can use, "box", "quartile", "point", "stick", None

````python
sns.violinplot(
    x = "time",
    y = "total_bill",
    data = var,
    order = ["Dinner", "Lunch"],
    inner = "quart",
)

plt.show()
````

We can make a single violin plot. But it only works on numerical data and not categorical

We can change the orientation of this violin by just using either "x" or "y" parameter.

````python
sns.violinplot(
    y = var["total_bill"],
)

plt.show()
````
