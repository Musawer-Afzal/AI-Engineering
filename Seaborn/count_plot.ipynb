What's the difference between countplot and barplot when they plot the same thing?

- Countplot plots the count of the number of records by category
- Barplots plots a value or metric for each category (by default, barplots plots the mean of a variable, by category)

````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
var = sns.load_dataset("tips")
var.head()
````

````python
sns.countplot(
    x = "sex",
    data = var,
)

plt.show()
````

````python
sns.barplot(
    x = "sex",
    y = "size",
    data = var
)

plt.show()
````

hue parameter, compare values according to a category

````python
sns.countplot(
    x = "sex",
    data = var,
    hue = "smoker",
)

plt.show()
````

y parameter, plot the graph horizontally

````python
sns.countplot(
    y = "sex",
    data = var,
    hue = "smoker",
)

plt.show()
````

palette parameter, change the complete color theme

````python
sns.countplot(
    x = "sex",
    data = var,
    hue = "smoker",
    palette = "bwr"
)

plt.show()
````
