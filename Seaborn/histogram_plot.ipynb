# Histogram Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
var = sns.load_dataset("penguins")
```

```python
var.head()
```

```python
sns.displot(
    var["bill_length_mm"],
)
```

The plt.show() function here is used to remove the printing line that tells us what is being printed. Seaborn can work without this show function

```python
sns.displot(
    var["bill_length_mm"],
)

plt.show()
```

```python
sns.displot(
    var["flipper_length_mm"],
)
```

bins parameter, we can add a fixed range

```python
sns.displot(
    var["flipper_length_mm"],
    bins = [170, 180, 190, 200, 210, 220, 230, 240]
)

plt.show()
```

```python
marks = [
    10, 12, 15, 18, 20, 22, 25, 28, 30, 35,
    40, 42, 45, 50, 55, 60, 65, 70, 75, 90
]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Automatic bin selection
sns.histplot(x=marks, ax=axes[0])
axes[0].set_title("Automatic bins")
    
# 10 bins
sns.histplot(x=marks, bins=10, ax=axes[1])
axes[1].set_title("bins=10")

# Custom bin edges
sns.histplot(
    x=marks,
    bins=[0, 20, 40, 60, 80, 100],
    ax=axes[2]
)
axes[2].set_title("Custom bin edges")

plt.tight_layout()
plt.show()
```

kde - Kernal density estimator, Shows movement patterns of values through a line

```python
sns.displot(
    var["flipper_length_mm"],
    bins = [170, 180, 190, 200, 210, 220, 230, 240],
    kde = True
)
```

rug parameter, make lines that shows the density of kde

```python
sns.displot(
    var["flipper_length_mm"],
    bins = [170, 180, 190, 200, 210, 220, 230, 240],
    kde = True,
    rug = True
)
```

color parameter, change the bars color

```python
sns.displot(
    var["flipper_length_mm"],
    bins = [170, 180, 190, 200, 210, 220, 230, 240],
    color = "red"
)

plt.show()
```

log scale, convert all the data to log scale

```python
sns.displot(
    var["flipper_length_mm"],
    # bins = [170, 180, 190, 200, 210, 220, 230, 240],
    kde = True,
    rug = True,
    color = "red",
    log_scale = True,
)

plt.show()
```
