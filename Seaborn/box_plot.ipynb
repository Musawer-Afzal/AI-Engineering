````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
var = sns.load_dataset('tips')
var
````

````python
sns.set(style = "whitegrid")
sns.boxplot(
    x = "day",
    y = "total_bill",
    data = var
)

plt.show()
````

hue parameter, divide the data according to a certain value

````python
sns.set(style = "whitegrid")
sns.boxplot(
    x = "day",
    y = "total_bill",
    data = var,
    hue = "sex",
)

plt.show()
````

color parameter

````python
sns.set(style = "whitegrid")
sns.boxplot(
    x = "day",
    y = "total_bill",
    data = var,
    hue = "sex",
    color = "brown"
)

plt.show()
````

order parameter, change the order of how the values are displayed

````python
sns.set(style = "whitegrid")
sns.boxplot(
    x = "day",
    y = "total_bill",
    data = var,
    hue = "sex",
    order = ["Fri", "Sun", "Thur", "Sat"]
)

plt.show()
````
