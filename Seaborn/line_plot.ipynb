````python
var = [1, 2, 3, 4, 5, 6, 7]
var_1 = [2, 3, 5, 7, 8, 4, 7]

x_1 = pd.DataFrame(
    {
        "var": var,
        "var_1": var_1
    }
)
````

Matplotlib graph

````python
plt.plot(var, var_1)
````

Seaborn Graph

We can either pass on the values directly to seabron's x and y parameter
```
sns.lineplot(x = value1, y = value2)
```
This will plot the values but there won't be any labels on x and y axis.

If we want to add labels then we need to use the **.xlabel, .ylabel** function.

If we do not want to use label function then we can assign values as strings to x and y and then pass on a separate variable called **data** where we'll pass the data we want to plot

Now the values in **x** and **y** will be used a labels for x and y axis
```
sns.lineplot(x = "value2", y = "value1", data = DataFrame)
```

````python
sns.lineplot(x = "var", y = "var_1", color = "r", data = x_1)
plt.show()
````
