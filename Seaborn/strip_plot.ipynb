````python
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
var = sns.load_dataset('tips')
var
````

````python
sns.stripplot(x = "day", y = "total_bill", data = var)

plt.show()
````

hue parameter, plot with regards to a middle value

````python
sns.stripplot(x = "day", y = "total_bill", data = var, hue = "sex")

plt.show()
````

palette parameter, change the color theme

````python
sns.stripplot(
    x = "day", 
    y = "total_bill", 
    data = var, 
    hue = "sex",
    palette = "rocket_r"
    )

plt.show()
````

linewidth parameter, change the width of edges of dots

````python
sns.stripplot(
    x = "day", 
    y = "total_bill", 
    data = var, 
    hue = "sex",
    linewidth = 3,
    )

plt.show()
````
