# MultiThreading and MultiProcessing

First we are going to run a normal function and notice the time

```python
import time
import threading
```

```python
def square(num):
    for i in range(len(num)):
        time.sleep(0.2)
        print(f"Square: {i*i}")

def cube(num):
    for i in range(len(num)):
        time.sleep(0.2)
        print(f"Cube: {i*i*i}")
```

```python
arr = [1, 4, 6, 7]

t = time.time()

square(arr)
cube(arr)
print("Done in ", time.time() - t, "seconds")
```
