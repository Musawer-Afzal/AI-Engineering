### Class and Instance Attributes

**Class Attributes** are the one which are same for every object we are trying to make

For example: If we are making a database for a college then the college name would be the same for every student so we won't need to declare them again and again for every student we create

````python
class Student:
    college_name = "Best College"
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("John", 30)
s2 = Student("Jane", 25)

print(s1.college_name)
print(s2.college_name)
print(s1.name)
print(s2.name)
print(s1.age)
print(s2.age)
````

**Instance Attributes** are the one which can be different for every object we create

Like name of the student and their age, address etc

**Name** and **Age** are instance Attributes here in this example

````python
class Student:
    college_name = "Best College"
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("John", 30)
s2 = Student("Jane", 25)

print(s1.college_name)
print(s2.college_name)
print(s1.name)
print(s2.name)
print(s1.age)
print(s2.age)
````

### Abstraction

Hiding the implementation details of a class and only showing the essential features to the user

````python
class Car:
    def __init__(self):
        self.acc = False
        self.brake = False
        self.clutch = False

    def start(self):
        self.clutch = True
        self.acc = True
        print("Car started")


c1 = Car()
print(f"Accelerator: {c1.acc}\n"
      f"Brake: {c1.brake}\n"
      f"Clutch: {c1.clutch}")

c1.start()
print("=" * 5, "Car Started", "=" * 5)
print(f"Accelerator: {c1.acc}\n"
      f"Brake: {c1.brake}\n"
      f"Clutch: {c1.clutch}")
````

````python
class Account:
    def __init__(self, balance, account_no):
        self.balance = balance
        self.account_no = account_no

    def debit(self, amount):
        self.balance -= amount
        print(f"Debited {amount} from account number {self.account_no}")
        print(f"New balance is: {self.balance}")

    def credit(self, amount):
        self.balance += amount
        print(f"Credited {amount} to account number {self.account_no}")
        print(f"New balance is: {self.balance}")


a1 = Account(1000, 1234)
a1.debit(500)
a1.credit(1000)
````

## Inheritance
When one class(child/derived) derives the properties and methods of another class(parent/base)

We have 3 types of Inheritance
- Singel Inheritance
- Multi-level inheritance
- Multiple inheritance

### Single Inheritance

````python
class Car:
    @staticmethod
    def start():
        print("Car Started...")

    @staticmethod
    def stop():
        print("Car Stopped...")

class ToyotaCar(Car):
    def __init__(self, name):
        self.name = name


car1 = ToyotaCar("Camry")
print(car1.name)
car1.start()
car1.stop()
````

### Multi-level Inheritance

A child class acts as a parent for another class

### Multiple Inheritance

A derived class inherits from More than one class

````python
class A:
    varA = "Welcome to class A"

class B:
    varB = "Welcome to class B"

class C(A, B):
    varC = "Welcome to class C"

c1 = C()
print(c1.varA, c1.varB, c1.varC)
````

### Super Method

````python
class Car:
    def __init__(self, type):
        self.type = type

    @staticmethod
    def start():
        print("Car Started...")

    @staticmethod
    def stop():
        print("Car Stopped...")


class ToyotaCar(Car):
    def __init__(self, brand, type):
        self.brand = brand
        super().__init__(type)
        super().start()



car1 = ToyotaCar("Toyata", "Camry")
print(car1.type)
````

### Class Method

A class method is bound to the class and recieves the class as an implicit first argument

Note - static method can't access or modify class state and generally for utility

```
class Student:
    @classmethod # decorator
    def college(cls):
        pass
```

We take the class as the first argument itself

````python
class Person:
    name = "anonymous"

    def changeName(self, name):
        self.name = name


p1 = Person()
p1.changeName("John")
print(p1.name)
print(Person.name)
````

The name remains the same for Person class

This is because self.name created a instance of the variable for this one object. The class value itself did not change

**There are different methods we can use to change the value of this attribute**

````python
class Person:
    name = "anonymous"

    def changeName(self, name):
        Person.name = name


p1 = Person()
p1.changeName("John")
print(p1.name)
print(Person.name)
````

````python
class Person:
    name = "anonymous"

    def changeName(self, name):
        self.__class__.name = "Doe"
        # self.__class__.name, Person.name


p1 = Person()
print(p1.name)
p1.changeName("John")
print(p1.name)
print(Person.name)
````

````python
class Person:
    name = "anonymous"

    @classmethod
    def changeName(cls, name):
        cls.name = name


p1 = Person()
print(p1.name)
p1.changeName("John")
print(p1.name)
print(Person.name)
````

All in all we have **3 types of methods**

- instance methods: Normal methods that uses the self argument and other parameters
- staticmethod: They do not take any argument and are created one time in a class when we create a object.
- class method: They take class as a first argument and are used to update value of the class itself

### Property Decorator

````python
class Student:
    def __init__(self, phy, chem, math):
        self.phy = phy
        self.chem = chem
        self.math = math
        self.percentage = 0

    def calcPercentage(self):
        self.percentage = str((self.phy + self.chem + self.math) / 3) + "%"
        print(self.percentage)


stu1 = Student(56, 87, 90)
stu1.calcPercentage()

# now lets say we wnat to change the marks of a student
stu1.phy = 66
print(stu1.phy)
# Marks are updated but if we check the percentage, it remains the same
stu1.calcPercentage()
````

## Polymorphism: Operator Overloading
When the same operator is allowed to have different meaning according to the context

**Operators and Dunder functions**

```
- a + b # addition          a.__add__(b)
- a - b # subtraction       a.__sub__(b)
- a * b # multiplication    a.__mul__(b)
- a / b # division          a.__truediv__(b)
- a % b # modulus           a.__mod__(b)
```

Every addition is dependent on the type of data

````python
print(1 + 1)
print("1" + "1")
print([1 , 2, 3] + [4, 5, 6])
````

````python
class Complex:
    def __init__(self, real, img):
        self.real = real
        self.img = img

    def showNumber(self):
        print(f"{self.real}i + {self.img}j")

    def add(self, num2):
        newReal = self.real + num2.real
        newImg = self.img + num2.img
        return Complex(newReal, newImg)


num1 = Complex(1, 3)
num1.showNumber()

num2 = Complex(4, 6)
num2.showNumber()

num3 = num1.add(num2)
num3.showNumber()
````

But what we want is that we can do:

````python
num3 = num1 + num2
````

And it should not throw any error

That is what the **Dunder** Functions are for

These functions already exist, we just have to give them the required logic to execute our problems

````python
class Complex:
    def __init__(self, real, img):
        self.real = real
        self.img = img

    def showNumber(self):
        print(f"{self.real}i + {self.img}j")

    def __add__(self, num2):
        newReal = self.real + num2.real
        newImg = self.img + num2.img
        return Complex(newReal, newImg)


num1 = Complex(1, 3)
num1.showNumber()

num2 = Complex(4, 6)
num2.showNumber()

num3 = num1 + num2
num3.showNumber()
````

We can use Similar logic for Subtraction too

````python
class Complex:
    def __init__(self, real, img):
        self.real = real
        self.img = img

    def showNumber(self):
        print(f"{self.real}i + {self.img}j")

    def __add__(self, num2):
        newReal = self.real + num2.real
        newImg = self.img + num2.img
        return Complex(newReal, newImg)

    def __sub__(self, num2):
            newReal = self.real - num2.real
            newImg = self.img - num2.img
            return Complex(newReal, newImg)


num1 = Complex(1, 3)
num1.showNumber()

num2 = Complex(4, 6)
num2.showNumber()

num3 = num1 + num2
num3.showNumber()

num3 = num1 - num2
num3.showNumber()
````

````python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

    def perimeter(self):
        return 2 * 3.14 * self.radius


c1 = Circle(10)
print(c1.area())
print(c1.perimeter())
````
