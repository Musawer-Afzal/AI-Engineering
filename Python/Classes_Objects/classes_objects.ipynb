### Class and Instance Attributes

**Class Attributes** are the one which are same for every object we are trying to make

For example: If we are making a database for a college then the college name would be the same for every student so we won't need to declare them again and again for every student we create

````python
class Student:
    college_name = "Best College"
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("John", 30)
s2 = Student("Jane", 25)

print(s1.college_name)
print(s2.college_name)
print(s1.name)
print(s2.name)
print(s1.age)
print(s2.age)
````

**Instance Attributes** are the one which can be different for every object we create

Like name of the student and their age, address etc

**Name** and **Age** are instance Attributes here in this example

````python
class Student:
    college_name = "Best College"
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("John", 30)
s2 = Student("Jane", 25)

print(s1.college_name)
print(s2.college_name)
print(s1.name)
print(s2.name)
print(s1.age)
print(s2.age)
````

### Methods are functions that are present inside a Class

````python
class Student:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print(f"Hello, {self.name}")


s1 = Student("John")
s1.greet()
````

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def calculate_average(self):
        return f"Hello, {self.name}. Your average is {sum(self.marks) / len(self.marks)}"


s1 = Student("John", [45, 76, 86])
s1.calculate_average()
````

## 2 Pillars of OOP

### Abstraction

Hiding the implementation details of a class and only showing the essential features to the user

````python
class Car:
    def __init__(self):
        self.acc = False
        self.brake = False
        self.clutch = False

    def start(self):
        self.clutch = True
        self.acc = True
        print("Car started")


c1 = Car()
print(f"Accelerator: {c1.acc}\n"
      f"Brake: {c1.brake}\n"
      f"Clutch: {c1.clutch}")

c1.start()
print("=" * 5, "Car Started", "=" * 5)
print(f"Accelerator: {c1.acc}\n"
      f"Brake: {c1.brake}\n"
      f"Clutch: {c1.clutch}")
````

### Practise

````python
class Account:
    def __init__(self, balance, account_no):
        self.balance = balance
        self.account_no = account_no

    def debit(self, amount):
        self.balance -= amount
        print(f"Debited {amount} from account number {self.account_no}")
        print(f"New balance is: {self.balance}")

    def credit(self, amount):
        self.balance += amount
        print(f"Credited {amount} to account number {self.account_no}")
        print(f"New balance is: {self.balance}")


a1 = Account(1000, 1234)
a1.debit(500)
a1.credit(1000)
````

### Private Attribte and Methods
**Conceptual Implementation in Python**:

Private attributes and methods are meant to be used only within the class and are not accessible from outside the class

````python
class Account:
    def __init__(self, acc_no, acc_pass):
        self.acc_no = acc_no
        self.__acc_pass = acc_pass

    def reset_password(self, new_pass):
        self.__acc_pass = new_pass

    def get_password(self):
        return self.__acc_pass

acc1 = Account("12345", "abcde")
print(acc1.acc_no)

acc1.reset_password("new_password")
print(acc1.acc_no, acc1.get_password())
````

## Inheritance
When one class(child/derived) derives the properties and methods of another class(parent/base)

We have 3 types of Inheritance
- Singel Inheritance
- Multi-level inheritance
- Multiple inheritance

### Single Inheritance

````python
class Car:
    @staticmethod
    def start():
        print("Car Started...")

    @staticmethod
    def stop():
        print("Car Stopped...")

class ToyotaCar(Car):
    def __init__(self, name):
        self.name = name


car1 = ToyotaCar("Camry")
print(car1.name)
car1.start()
car1.stop()
````

### Multi-level Inheritance

A child class acts as a parent for another class

````python
class Car:
    @staticmethod
    def start():
        print("Car Started...")

    @staticmethod
    def stop():
        print("Car Stopped...")

class ToyotaCar(Car):
    def __init__(self, brand):
        self.brand = brand

class Fortuner(ToyotaCar):
    def __init__(self, type):
        self.type = type



car1 = ToyotaCar("Toyata")
print(car1.brand)
car1.start()
car1.stop()
car2 = Fortuner("petrol")
car2.start()
````

### Multiple Inheritance

A derived class inherits from More than one class

````python
class A:
    varA = "Welcome to class A"

class B:
    varB = "Welcome to class B"

class C(A, B):
    varC = "Welcome to class C"

c1 = C()
print(c1.varA, c1.varB, c1.varC)
````
