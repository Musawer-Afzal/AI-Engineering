Classes are blueprint for objects

**__init__** Function

````python
class Students:
    def __init__(self, name):
        self.name = name
        print("Student declared")


s1 = Students("John")
````

The **self** parameter is a reference to the current instance of the class, and is used to access variables that belongs to the class

### Class and Instance Attributes

**Class Attributes** are the one which are same for every object we are trying to make

For example: If we are making a database for a college then the college name would be the same for every student so we won't need to declare them again and again for every student we create

````python
class Student:
    college_name = "Best College"
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("John", 30)
s2 = Student("Jane", 25)

print(s1.college_name)
print(s2.college_name)
print(s1.name)
print(s2.name)
print(s1.age)
print(s2.age)
````

**Instance Attributes** are the one which can be different for every object we create

Like name of the student and their age, address etc

**Name** and **Age** are instance Attributes here in this example

````python
class Student:
    college_name = "Best College"
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("John", 30)
s2 = Student("Jane", 25)

print(s1.college_name)
print(s2.college_name)
print(s1.name)
print(s2.name)
print(s1.age)
print(s2.age)
````

### Methods are functions that are present inside a Class

````python
class Student:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print(f"Hello, {self.name}")


s1 = Student("John")
s1.greet()
````

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def calculate_average(self):
        return f"Hello, {self.name}. Your average is {sum(self.marks) / len(self.marks)}"


s1 = Student("John", [45, 76, 86])
s1.calculate_average()
````

### Static Methods

Methods that do not use self parameter

**Syntax:**

The term **@staticmethod** is called a **Decorator**

**Decorators:**
- Decorators allow us to wrap another function in order to extend the behaviour of the wrapped function, without permanently modifying it.

## 2 Pillars of OOP

### Abstraction

Hiding the implementation details of a class and only showing the essential features to the user

````python
class Car:
    def __init__(self):
        self.acc = False
        self.brake = False
        self.clutch = False

    def start(self):
        self.clutch = True
        self.acc = True
        print("Car started")


c1 = Car()
print(f"Accelerator: {c1.acc}\n"
      f"Brake: {c1.brake}\n"
      f"Clutch: {c1.clutch}")

c1.start()
print("=" * 5, "Car Started", "=" * 5)
print(f"Accelerator: {c1.acc}\n"
      f"Brake: {c1.brake}\n"
      f"Clutch: {c1.clutch}")
````

### Encapsulation
Wrapping data and function into a single unit(object)

Creating objects, and making classes up until now is all Encapsulation

### Practise

````python
class Account:
    def __init__(self, balance, account_no):
        self.balance = balance
        self.account_no = account_no

    def debit(self, amount):
        self.balance -= amount
        print(f"Debited {amount} from account number {self.account_no}")
        print(f"New balance is: {self.balance}")

    def credit(self, amount):
        self.balance += amount
        print(f"Credited {amount} to account number {self.account_no}")
        print(f"New balance is: {self.balance}")


a1 = Account(1000, 1234)
a1.debit(500)
a1.credit(1000)
````
