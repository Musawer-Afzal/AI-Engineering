### Exceptions
If things go wrong during the execution of the program(runtime). It generally happens when something unforseen has happened.
- Exceptions are raised by Python runtime
- You have to takle this on the fly

**Example**
- Memory Overflow
- Divide by 0 -> logical error
- Database Error

```python
# Lets create a file
with open("sample.txt", "w") as f:
    f.write("hello")
```

```python
try:
    with open("sample.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("File not found")
```

```python
try:
    f = open("sample.txt", "r")
    m = 4
    print(m)
    print(5/2)
    L = [1, 2, 3]
    print(L[1])
except FileNotFoundError:
    print("File not Found")
except NameError:
    print("Variable not defined")
except ZeroDivisionError:
    print("Cannot divide by zero")
except Exception as e:
    print(e)
else:
    print(f.read())
    print("File is present")
finally:
    f.close()
```

## Creating Custom Exceptions

```python
class MyException(Exception):
    def __init__(self, msg):
        print(msg)

class Bank:
    def __init__(self, balance):
        self.balance = balance
    
    def withdraw(self, amount):
        if amount < 0:
            raise MyException('amount cannot be -ve')
        if self.balance < amount:
            raise MyException("Insufficient balance")
        self.balance -= amount

bankAccount = Bank(2000)
try:
    bankAccount.withdraw(3000)
except MyException as e:
    pass
else:
    print(bankAccount.balance)
```
