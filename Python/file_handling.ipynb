### How file I/O is done in most programming languages
- Open a file
- Read/Write data
- Close the file

Writing to a file

```python
# case 1 - if the file is not present
try:
    f = open("sample.txt", "w")
    f.write("Hello World")
    f.close()
except FileNotFoundError:
    print("File not found")
```

```python
# Write multiple strings
try:
    f = open("sample.txt", "w")
    f.write("Hello to the new world")
    f.write("\nLets move to the next line")
    f.close()
except FileNotFoundError:
    print("File not found")
```

```python
# case 2 - if the file is present
try:
    f = open("sample.txt", "w")
    f.write("File is present")
    f.close()
except FileNotFoundError:
    print("File not found")
else:
    print("File is present")
```

How exactly open works

- When we write **f - open("file_name", "type_access")**, the file is loaded from harddisk to RAM
- The file is added to the buffer zone where any changes to the file are done
- After we call **f.close()**, the file is removed from RAM and put back in Harddisk

```python
# Problem with w mode
# Introducing append mode
try:
    f = open("sample.txt", "a")
    f.write("\nI am fine")
    f.close()
except FileNotFoundError:
    print("File not found")
else:
    print("File is present")
```

```python
# Write multiple lines
L = ["hello", "\nhi", "\nhow are you?", "\nI am fine"]
try:
    f = open("sample.txt", "w")
    f.writelines(L)
    f.close
except FileNotFoundError:
    print("File not found")
else:
    print("File is present")
```

### Using Context Manager (With keyword)
- It's a good idea to close a file after usage as it will free up the resources, meaning free up the RAM
- If we don't close it, garbage collector will close it
- with keyboard close the file as soon as the usage is over

```python
# with
with open("sample.txt", "w") as f:
    f.write("Superman")
```

```python
# try f.read()
with open("sample.txt", "r") as f:
    print(f.read())
```

### Moving within a file -> 10 char than 10 char

```python
with open("sample.txt", "r") as f:
    print(f.read(5))
    print(f.read(5))
```

Benifit of using buffer, as the above example printed 5 characters first and then the next 5 characters. This way we can deal with way bigger files and deal with them in iterations

```python
big_L = [f"Hello world{i}\n" for i in range(1000)]
with open("big.txt", "w") as f:
    f.writelines(big_L)
```

```python
with open("big.txt", "r") as f:
    chunk_size = 100
    while len(f.read(chunk_size)) > 0:
        print(f.read(chunk_size))
        f.read(chunk_size)
```

Seek and tell function

**Seek** is used to tell where the read function is. It gives us the index where the read function is at.

**Tell** is used to move the index at which the read function is going to print at, anywhere else. Before this, read keeps on moving forward and keeps track of what it has already printed and what it is going to print next

```python
with open("sample.txt", "r") as f:
    print(f.read(10))
    print(f.tell()) # returns the current position
    f.seek(0) # move the cursor to the beginning
    print(f.read(10)) # read the first 10 characters again
```

### Serialization and Deserialization
- **Serialization** - a process of converting python data types to JSON format
- **Deserialization** - process of converting JSON to python data types

Serialization using python modules

### List
**Syntax**

```python
import json

L = [1, 2, 3, 4]

with open("demo.json", "w") as f:
    json.dump(L, f)
```

```python
with open("demo.json", "w") as f:
    json.dump(d, f, indent = 4) # how many indentations before every key value pair
```

### Deserialization

```python
with open("demo.json", "r") as f:
    file = json.load(f)
    print(file)
    print(type(file))
```

## Serializing and Deserializing custom objects

```python
class Person:
    def __init__(self, fname, lname, age, gender):
        self.fname = fname
        self.lname = lname
        self.age = age
        self.gender = gender
```

```python
person = Person("John", "Doe", 30, "male")
```

```python
# This will cause an error
with open("demo.json", "w") as f:
    json.dump(person, f)
```

This is how we fix it

We can save it as a string

```python
def show_object(person):
    if isinstance(person, Person):
        return "{} {} age -> {} gender -> {}".format(person.fname, person.lname, person.age, person.gender)


with open("demo.json", "w") as f:
    json.dump(person, f, default = show_object)
```

We can also save it as a dictionary

```python
def show_object(person):
    if isinstance(person, Person):
        return {
            "fname": person.fname,
            "lname": person.lname,
            "age": person.age,
            "gender": person.gender
        }


with open("demo.json", "w") as f:
    json.dump(person, f, default = show_object, indent = 4)
```

Deserialization

```python
with open("demo.json", "r") as f:
    data = json.load(f)
    print(data)
    print(type(data))
```
