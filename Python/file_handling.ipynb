# File Handling in Python

Types of data used for I/O
- Text - '12345' as a sequence of unicode chars
- Binary - 12345 as a sequence of bytes of its binary equivalent

Hence there are 2 file types to deal with
- Text files - All program filea re text files
- Binary files - Images, videos, music, exe files

### How file I/O is done in most programming languages
- Open a file
- Read/Write data
- Close the file

Writing to a file

```python
# case 1 - if the file is not present
try:
    f = open("sample.txt", "w")
    f.write("Hello World")
    f.close()
except FileNotFoundError:
    print("File not found")
```

```python
# Write multiple strings
try:
    f = open("sample.txt", "w")
    f.write("Hello to the new world")
    f.write("\nLets move to the next line")
    f.close()
except FileNotFoundError:
    print("File not found")
```

```python
# case 2 - if the file is present
try:
    f = open("sample.txt", "w")
    f.write("File is present")
    f.close()
except FileNotFoundError:
    print("File not found")
else:
    print("File is present")
```

How exactly open works

- When we write **f - open("file_name", "type_access")**, the file is loaded from harddisk to RAM
- The file is added to the buffer zone where any changes to the file are done
- After we call **f.close()**, the file is removed from RAM and put back in Harddisk

```python
# Problem with w mode
# Introducing append mode
try:
    f = open("sample.txt", "a")
    f.write("\nI am fine")
    f.close()
except FileNotFoundError:
    print("File not found")
else:
    print("File is present")
```

```python
# Write multiple lines
L = ["hello", "\nhi", "\nhow are you?", "\nI am fine"]
try:
    f = open("sample.txt", "w")
    f.writelines(L)
    f.close
except FileNotFoundError:
    print("File not found")
else:
    print("File is present")
```

Reading from files

```python
# using read()
try:
    f = open("sample.txt", "r")
    s = f.read()
    f.close()
except FileNotFoundError:
    print("File not found")
else:
    print("File is present")

print(s)
```

Reading upto n characters

```python
try:
    f = open("sample.txt", "r")
    s = f.read(15)
    f.close()
except FileNotFoundError:
    print("file not found")
else:
    print("file found")

print(s)
```

```python
try:
    f = open("sample.txt", "r")
    s1 = f.readline()
    s2 = f.readline()
    f.close()
except FileNotFoundError:
    print("File not found")
else:
    print("File is present")

print(s1, end = '')
print(s2, end = '')
```

Reading entire file using readline

```python
try:
    f = open("sample.txt", "r")
    while(f.readline() != ''):
        print(f.readline(), end = '')
    f.close()
except FileNotFoundError:
    print("File not found")
```
