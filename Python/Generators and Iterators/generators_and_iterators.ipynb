## Iterators

```python
num = [1, 2, 3]
for i in num:
    print(i)
```

**What is a iterator?**

An iterator is an object that allows the programmer to traverse through a sequence of data without having to store the entire data in the memory

```python
L = [x for x in range(1, 100001)]
# for i in L:
#     print(i * 2)
```

Now lets see the size that it occupies in the memory

```python
import sys
sys.getsizeof(L) / 1024
```

Now lets check another approach

```python
x = range(1, 100001)
sys.getsizeof(x) / 1024
```

Now this is the difference

When we made the list **L** all of the integers from 1 to 100000 was created and stored in the memory at the same time. 

When we used **range**, it created a iterator that starts from 1 and picks up 1 from memory, multiply by 2 and then removed it. Then did the same for 2 and so on

### What is a Iterable?
Iterable is an object, which one can iterate over.

It generated an iterator when passed to **iter()** method

**For Example:** The list L that we made is an iterable. So is the x that we made. Meaning that everything that can be iterated is a iterable. OR anything we can run a loop on.

```python
# Example

L = [1, 2, 3]
print(type(L))

# L is an iterable
print(iter(L))

# iter(L)  --> iterator  
```

### Points to Remember
1. Every iterator is also a iterable
    - You can perform iteration on a iterator
2. Not all iterables are iterators
    - For Example: The list L that we made is a iterable but the condition that a iterator DOES NOT store all the data in the memory is not satisfied by L so it is not a iterator.

### Trick
- Every iterable has an iter function
- Every iterator has both iter function as well as a next function

```python
a = 2
a
# for i in a:
#     print(a) # Error: a is not iterable
```

A better method to find if an object is iterable is **dir()** function

```python
print(type(dir(a)), "\n")

if "__iter__" in dir(a):
    print("Iterable")
else:
    print("Not Iterable")
```

```python
T = (1, 2, 3)
if "__iter__" in dir(T):
    print("Iterable")
else:
    print("Not Iterable")
```
