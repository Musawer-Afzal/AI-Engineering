## Iterators

```python
num = [1, 2, 3]
for i in num:
    print(i)
```

**What is a iterator?**

An iterator is an object that allows the programmer to traverse through a sequence of data without having to store the entire data in the memory

```python
L = [x for x in range(1, 100001)]
# for i in L:
#     print(i * 2)
```

Now lets see the size that it occupies in the memory

```python
import sys
sys.getsizeof(L) / 1024
```

Now lets check another approach

```python
x = range(1, 100001)
sys.getsizeof(x) / 1024
```

Now this is the difference

When we made the list **L** all of the integers from 1 to 100000 was created and stored in the memory at the same time. 

When we used **range**, it created a iterator that starts from 1 and picks up 1 from memory, multiply by 2 and then removed it. Then did the same for 2 and so on

### What is a Iterable?
Iterable is an object, which one can iterate over.

It generated an iterator when passed to **iter()** method

**For Example:** The list L that we made is an iterable. So is the x that we made. Meaning that everything that can be iterated is a iterable. OR anything we can run a loop on.

```python
# Example

L = [1, 2, 3]
print(type(L))

# L is an iterable
print(iter(L))

# iter(L)  --> iterator  
```

### Points to Remember
1. Every iterator is also a iterable
    - You can perform iteration on a iterator
2. Not all iterables are iterators
    - For Example: The list L that we made is a iterable but the condition that a iterator DOES NOT store all the data in the memory is not satisfied by L so it is not a iterator.

### Trick
- Every iterable has an iter function
- Every iterator has both iter function as well as a next function

A better method to find if an object is iterable is **dir()** function

```python
print(type(dir(a)), "\n")

if "__iter__" in dir(a):
    print("Iterable")
else:
    print("Not Iterable")
```

```python
T = (1, 2, 3)
if "__iter__" in dir(T):
    print("Iterable")
else:
    print("Not Iterable")
```

Now how to find if an object is **Iterator** or not
- Run **dir()** and see if the fucntions **__Iter__** and **__next__** are available or not. If yes then they are iterator

```python
L = [1, 2, 3]
if "__iter__" and "__next__" in dir(L):
    print("L is an Iterator")
else:
    print("Not an iterator")
```

We discussed that we can find the iterator of a object using the **iter()** method

### Understading how the for loop works

```python
nums = [1, 2, 3]
for i in num:
    print(i)
```

There is a step by step process going on when we run this loop

```python
# Step 1: Fetch the iterator
iter_num = iter(num)
print(iter_num)

# Step 2: Calls the next function
print(next(iter_num))
print(next(iter_num))
print(next(iter_num))
print(next(iter_num))

# Step 3:  
```

```python
def my_for_loop(iterable):
    iterator = iter(iterable)
    while True:
        try:
            print(next(iterator))
        except StopIteration:
            print("Iterable ended")
            break    
```

```python
a = [1, 2, 3]
b = range(1, 11)
c = (1 ,2, 3)
d = {1, 2, 3}
e = {0: 1, 1: 1}

my_for_loop(a)
my_for_loop(b)
my_for_loop(c)
my_for_loop(d)
my_for_loop(e)
```

### Let's create our own range function

```python
class MyRange:
    def __init__(self, start, end):
        self.start = start
        self.end = end
    
    def __iter__(self):
        return MyRangeIterator(self)
```

```python
class MyRangeIterator:
    def __init__(self, iterable_obj):
        self.iterable = iterable_obj

    def __iter__(self):
        return self 

    def __next__(self):
        if self.iterable.start >= self.iterable.end:
            raise StopIteration
        
        current = self.iterable.start
        self.iterable.start += 1
        return current
```

```python
for i in MyRange(1, 11):
    print(i)
```

# Generators

Python generators are a simple way of creating iterators

```python
def gen_demo():
    yield "first statement"
    yield "second statement"
    yield "third statement"
```

```python
gen = gen_demo()
print(gen)
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
gen = gen_demo()
for i in gen:
    print(i)
```

```python
def square(num):
    for i in range(1, num):
        yield i**2

# We can print it normally and then if we run the loop, the rest will be printed
gen = square(10)
print(next(gen))
print(next(gen))
print(next(gen))
for i in gen:
    print(i)
```

### Difference between yield and return

**Normal Function:** A normal function runs one time and then is removed from the memory once its job is done

**Generator:** A generator when used, does its job one time, then returns. When called again, it continues from where it left off. It remembers where it stopped

## Range function using Generator

```python
def my_range(start, end):
    while start < end:
        yield start
        start += 1

gen = my_range(1, 11)
for i in gen:
    print(i)
```

### Benefits of using Generators

### 1. Ease of Implementation

Iterator VS Generator

```python
class MyRange:
    def __init__(self, start, end):
        self.start = start
        self.end = end
    
    def __iter__(self):
        return MyRangeIterator(self)

class MyRangeIterator:
    def __init__(self, iterable_obj):
        self.iterable = iterable_obj

    def __iter__(self):
        return self 

    def __next__(self):
        if self.iterable.start >= self.iterable.end:
            raise StopIteration
        
        current = self.iterable.start
        self.iterable.start += 1
        return current
```

```python
def my_range(start, end):
    while start < end:
        yield start
        start += 1

gen = my_range(1, 11)
for i in gen:
    print(i)
```

### 2. Memory Efficient

```python
import sys
L = [x for x in range(1, 100000)]
gen = (x for x in range(1, 100000))

print(f"List Size: {sys.getsizeof(L)}")
print(f"Generator Size: {sys.getsizeof(gen)}")
```

### 3. Representing Infinite Streams

```python
def all_even():
    n = 0
    while True:
        yield n
        n += 2

even_num = all_even()
next(even_num)
next(even_num)
```

### 4. Chaining Generators

```python
def fibonacci_numbers(nums):
    x , y = 0, 1
    for _ in range(nums):
        x, y = y, x + y
        yield x

def square(nums):
    for num in nums:
        yield num ** 2

print(sum(square(fibonacci_numbers(10))))
```
