## Bubble Sort

```python
L = [1, 435, 7, 8,5,588, 65, 79, 555, 88658, 96, 5,3, 879]
L1 = [1, 2, 3, 4, 5, 6]
len(L)
```

```python
for i in range(len(L)):
    isSwap = False
    for j in range(len(L) - 1):
        if(L[j] > L[j + 1]):
            L[j], L[j + 1] = L[j + 1], L[j]
            isSwap = True
    if (isSwap == False):
        break
        
print(L)
```

## Selection Sort

```python
L = [1, 435, 7, 8,5,588, 65, 79, 555, 88658, 96, 5,3, 879]
```

```python
for i in range(len(L) - 1):
    smallest_index = i
    for j in range(i + 1, len(L)):
        if(L[j] < L[smallest_index]):
            smallest_index = j
    L[i], L[smallest_index] = L[smallest_index], L[i]

print(L)
```

## Insertion Sort

```python
L = [1, 435, 7, 8,5,588, 65, 79, 555, 88658, 96, 5,3, 879]
```

```python
for i in range(1, len(L)):
    curr = i
    prev = i - 1
    while(prev >= 0 and L[prev] > L[curr]):
        L[prev + 1] = L[prev]
        prev -= 1
    L[prev + 1] = L[curr]

print(L)
```
