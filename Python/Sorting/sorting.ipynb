## Bubble Sort

```python
L = [1, 435, 7, 8,5,588, 65, 79, 555, 88658, 96, 5,3, 879]
L1 = [1, 2, 3, 4, 5, 6]
len(L)
```

```python
for i in range(len(L)):
    isSwap = False
    for j in range(len(L) - 1):
        if(L[j] > L[j + 1]):
            L[j], L[j + 1] = L[j + 1], L[j]
            isSwap = True
    if (isSwap == False):
        break
        
print(L)
```
