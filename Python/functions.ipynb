### Higher Order Functions

```python
def sum(num_list):
    even_sum = 0
    odd_sum = 0
    div3_sum = 0

    for i in num_list:
        if i % 2 == 0:
            even_sum += i
        if i % 2 != 0:
            odd_sum += i
        if i % 3 == 0:
            div3_sum += i

    return even_sum, odd_sum, div3_sum
```

```python
num_list = [23,45, 34, 21, 45, 67 ,7 ,5, 8,98]
even, odd, div3 = sum(num_list)
```

```python
print(f"Even sum: {even}\nOdd sum: {odd}\nDiv3 sum: {div3}")
```

```python
x = lambda x: x % 2 == 0
y = lambda x: x % 2 != 0
z = lambda x: x % 3 == 0
```

```python
def return_sum(func, n):
    result = 0

    for i in n:
        if func(i):
            result += i

    return result
```

```python
print(f"Even sum: {return_sum(x, num_list)}\nOdd sum: {return_sum(y, num_list)}\nDiv3 sum: {return_sum(z, num_list)}")
```

Lets see some built in higher order functions in Python
* Map
* Filter
* Reduce

1. Map
- Map requires two inputs, a function and a iterable, like a list

```python
L = [1, 2, 3, 4, 5, 6, 7]
res = map(lambda x: x * x, L)
res
```

```python
list(res)
```

```python
list(map(lambda x: x % 2 == 0, L))
```

```python
students = [
    {
        "name": "John",
        "father name": "Smith",
        "Address": "123 Hill Street",
    },
    {
        "name": "Jane",
        "father name": "Williams",
        "Address": "456 Main Street",
    },
    {
        "name": "Bob",
        "father name": "Brown",
        "Address": "789 Oak Avenue",
    },
]
```

```python
list(map(lambda student: student["name"], students))
```

2. Filter
- It is used to filter out the output according to the condition we pass

```python
num_list
```

```python
list(filter(lambda x: x > 20, num_list))
```

```python
fruits = ["apple", "banana", "Orange", "Grape"]
list(filter(lambda fruit: "e" in fruit, fruits))
```

3. Reduce
- It reduces the given input to one output unlike map and filter what gives us output in a list

```python
from functools import reduce
```

```python
num_list
```

```python
reduce(lambda x, y: x + y, num_list)
```

```python
reduce(lambda x, y: x if x > y else y, num_list)
```

### Dictionary Comprehension

```python
D = {
    "name": "John",
    "gender": "Male",
    "age": 30,
    "city": "New York",
    "country": "USA",
}
```

```python
D.items()
```

```python
D1 = [(key, value) for key, value in D.items() if len(key) > 4]
D1
```

```python
D2 = {key: value for key, value in D.items() if len(key) > 4}
D2
```
