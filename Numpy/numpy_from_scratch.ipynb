### Aggregate functions

- np.sum()
- np.mean()
- np.min()
- np.max()
- np.std(), standared deviation
- np.var(), Variance

```python
# Aggregate functions
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
print("Sum: ", np.sum(arr1))
print("Mean: ", np.mean(arr1))
print("Max: ", np.max(arr1))
print("Min: ", np.min(arr1))
print("Standard Deviation: ", np.std(arr1))
print("Variance: ", np.var(arr1))
print("Median: ", np.median(arr1))
print("Percentile: ", np.percentile(arr1, 50))
print("Sort: ", np.sort(arr1))
print("Reverse Sort: ", np.sort(arr1)[::-1])
print("Sort Axis: ", np.sort(arr1, axis = 0))
print("Reverse Sort Axis: ", np.sort(arr1, axis = 0)[::-1])
print("Sort Axis 1: ", np.sort(arr1, axis = 1))
print("Reverse Sort Axis 1: ", np.sort(arr1, axis = 1)[::-1])
```
