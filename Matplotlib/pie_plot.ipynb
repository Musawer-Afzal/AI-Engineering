# Pie Plot

```python
import matplotlib.pyplot as plt
```

Explode parameter, make a certain label prominent

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    colors = c,)
plt.show()
```

autopct, Show percentage of every value

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    # colors = c,
    autopct = "%0.1f%%" # the %0.1f%% means 1 decimal place, and %0.2f%% means 2 decimal places
    )
plt.show()
```

Shadow parameter

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    autopct = "%0.1f%%",
    shadow = True,
    )
plt.show()
```

radius, increase the size of pie

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    autopct = "%0.1f%%",
    shadow = True,
    radius = 1.5,
    )
plt.show()
```
