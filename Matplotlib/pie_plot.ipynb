# Pie Plot

```python
import matplotlib.pyplot as plt
```

Explode parameter, make a certain label prominent

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    colors = c,)
plt.show()
```

autopct, Show percentage of every value

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    # colors = c,
    autopct = "%0.1f%%" # the %0.1f%% means 1 decimal place, and %0.2f%% means 2 decimal places
    )
plt.show()
```

Shadow parameter

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    autopct = "%0.1f%%",
    shadow = True,
    )
plt.show()
```

radius, increase the size of pie

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    autopct = "%0.1f%%",
    shadow = True,
    radius = 1.5,
    )
plt.show()
```

labeldistance, create distance between the labels and pie chart

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    autopct = "%0.1f%%",
    shadow = True,
    labeldistance = 1.5,
    )
plt.show()
```

startangle parameter, change the starting angle of the pie

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    autopct = "%0.1f%%",
    shadow = True,
    startangle = 90,
    )
plt.show()
```

textprops parameter, change the fontsize of text. It takes values in the form of dictionary

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    autopct = "%0.1f%%",
    shadow = True,
    textprops = {"fontsize": 15},
    )
plt.show()
```

counterclock, change the pie's orientation to counterclockwise

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    autopct = "%0.1f%%",
    shadow = True,
    counterclock = False,
    )
plt.show()
```

wedgeprops parameter, We can change the wedge thickness, width, edgecolor etc

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    autopct = "%0.1f%%",
    shadow = True,
    wedgeprops = {
        "width": 0.6, # The distance from the middle. The hole that is created
        "linewidth": 5, # The width of the edge
        "edgecolor": "magenta"},
    )
plt.show()
```

rotate label, rotate the labels in the chart

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    autopct = "%0.1f%%",
    shadow = True,
    rotatelabels = True,
    )
plt.show()
```

plt.legend, Show all the labels used

```python
x = [10, 20,30, 40]
y = ["c", "c++", "Java", "Python"]
ex = [0.3, 0.0, 0.0, 0.0]
c = ["red", "yellow", "green", "blue"]

plt.pie(
    x, 
    labels = y,
    explode = ex,
    autopct = "%0.1f%%",
    shadow = True,
    )

plt.legend(loc = 1)
plt.title("Languages Popularity", fontsize = 20)
plt.show()
```
