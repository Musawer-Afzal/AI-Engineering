# Stem Plot

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5, 6]
y = [2, 2, 5, 6, 4, 3]

plt.stem(x, y)

plt.show()
```

linefmt parameter, to change the type of line

```python
x = [1, 2, 3, 4, 5, 6]
y = [2, 2, 5, 6, 4, 3]

plt.stem(x, y, linefmt = ":")

plt.show()
```

```python
x = [1, 2, 3, 4, 5, 6]
y = [2, 2, 5, 6, 4, 3]

plt.stem(
    x, 
    y, 
    linefmt = ":", 
    markerfmt = "r+", # "ro" for red circle etc
    )

plt.show()
```

bottom parameter, what value does the bottom line should start from

```python
x = [1, 2, 3, 4, 5, 6]
y = [2, 2, 5, 6, 4, 3]

plt.stem(x, y, bottom = 4)

plt.show()
```
