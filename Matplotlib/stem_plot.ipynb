# Stem Plot

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5, 6]
y = [2, 2, 5, 6, 4, 3]

plt.stem(x, y)

plt.show()
```

linefmt parameter, to change the type of line

```python
x = [1, 2, 3, 4, 5, 6]
y = [2, 2, 5, 6, 4, 3]

plt.stem(x, y, linefmt = ":")

plt.show()
```

```python
x = [1, 2, 3, 4, 5, 6]
y = [2, 2, 5, 6, 4, 3]

plt.stem(x, y, linefmt = ":", markerfmt = "r+")

plt.show()
```
