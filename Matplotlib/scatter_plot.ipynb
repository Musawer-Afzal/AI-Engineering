### Scatter Plot

```python
import matplotlib.pyplot as plt
```

```python
day = [1, 2, 3, 4, 5, 6, 7]
number = [2, 4, 5, 2, 5, 7, 8]

plt.scatter(day, number, color = "red")

plt.title("Scatter Plot", fontsize = 20)
plt.xlabel("Day", fontsize = 15)
plt.ylabel("Number", fontsize = 15)
plt.show()
```

How to use different colors

```python
day = [1, 2, 3, 4, 5, 6, 7]
number = [2, 4, 5, 2, 5, 7, 8]

color = ["blue", "green", "red", "yellow", "black", "orange", "pink"]
sizes = [100, 585, 235, 55, 675, 455, 995]
plt.scatter(day, 
            number, 
            color = color, 
            s = sizes, # Declare sizes of individual or all the points
            # alpha = 0.4, # Change the transparency of the points
            marker = "*", # Change the shape of points
            edgecolor = "green", # Change the color of edges of points
            linewidth = 1 # Change the width of edges of points
            )
            
plt.title("Scatter Plot", fontsize = 20)
plt.xlabel("Day", fontsize = 15)
plt.ylabel("Number", fontsize = 15)
plt.show()  
```
