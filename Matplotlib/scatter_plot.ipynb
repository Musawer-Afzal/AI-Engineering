### Scatter Plot

```python
import matplotlib.pyplot as plt
```

```python
day = [1, 2, 3, 4, 5, 6, 7]
number = [2, 4, 5, 2, 5, 7, 8]

plt.scatter(day, number, color = "red")

plt.title("Scatter Plot", fontsize = 20)
plt.xlabel("Day", fontsize = 15)
plt.ylabel("Number", fontsize = 15)
plt.show()
```

How to use different colors

```python
day = [1, 2, 3, 4, 5, 6, 7]
number = [2, 4, 5, 2, 5, 7, 8]

color = ["blue", "green", "red", "yellow", "black", "orange", "pink"]
sizes = [100, 585, 235, 55, 675, 455, 995]
plt.scatter(day, 
            number, 
            color = color, 
            s = sizes, # Declare sizes of individual or all the points
            # alpha = 0.4, # Change the transparency of the points
            marker = "*", # Change the shape of points
            edgecolor = "green", # Change the color of edges of points
            linewidth = 1 # Change the width of edges of points
            )
            
plt.title("Scatter Plot", fontsize = 20)
plt.xlabel("Day", fontsize = 15)
plt.ylabel("Number", fontsize = 15)
plt.show()  
```

We can use code to declare the color of points and then select from a variety of different themes of colors to use

```python
day = [1, 2, 3, 4, 5, 6, 7]
number = [2, 4, 5, 2, 5, 7, 8]

colors = [10, 49, 30, 29, 56, 20, 30] # The colors range from 0 to 100
sizes = [100, 585, 235, 55, 675, 455, 995]
plt.scatter(day, 
            number, 
            c = colors, 
            s = sizes, # Declare sizes of individual or all the points
            cmap = "autumn", # Change the color map, "viridus", "BrBG", "autumn", "coolwarm" etc
            alpha = 0.6
            )

t = plt.colorbar()  # This will show me the range of colors i am using
t.set_label("Colors", fontsize = 15) # Change the label of the color bar itself
plt.title("Scatter Plot", fontsize = 20)
plt.xlabel("Day", fontsize = 15)
plt.ylabel("Number", fontsize = 15)
plt.show()  
```
