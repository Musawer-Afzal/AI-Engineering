# Fill Between Plot

```python
import matplotlib.pyplot as plt
import numpy as np
```

You can either pass complete variable, e.g "x" and "y" here or you can pass some values

```python
x = [1, 2, 3, 4, 5]
y = [1, 2, 3, 4, 5]


plt.plot(x, y, color = "r")
plt.fill_between(x = [2, 4], y1 = 2, y2 = 4)



plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

color parameter, change the color of area

```python
x = [1, 2, 3, 4, 5]
y = [1, 2, 3, 4, 5]


plt.plot(x, y, color = "r")
plt.fill_between(x = [2, 4], y1 = 2, y2 = 4, color = "magenta")



plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

where parameter, if we do not want to specify limit of x and y directly, we can use where to do that.

- Also **where** parameter dose not works with list so we need to convert the values to array using numpy

```python
x = [1, 2, 3, 4, 5]
y = [1, 2, 3, 4, 5]


plt.plot(x, y, color = "r")
plt.fill_between(x, y, where = [x >= 2] & [x <= 4])



plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([1, 2, 3, 4, 5])


plt.plot(x, y, color = "r")
plt.fill_between(x, y, where = (x >= 2) & (x <= 4))



plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```
