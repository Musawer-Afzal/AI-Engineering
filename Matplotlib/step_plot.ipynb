# Step Plot

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.step(x, y)


plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.step(x, y)

plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.show()
```

marker parameter, mark the point where intersection meet

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.step(x, y, marker = "o")

plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.show()
```
