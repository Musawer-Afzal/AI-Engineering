# Step Plot

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.step(x, y)


plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.step(x, y)

plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.show()
```

marker parameter, mark the point where intersection meet

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.step(x, y, marker = "o")

plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.show()
```

color parameter, change the color of lines

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.step(x, y, color = "magenta")

plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.show()
```

ms parameter, change marker size

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.step(x, y, marker = "o", ms = "10")

plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.show()
```

mfc parameter, change marker color

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.step(x, y, marker = "o", mfc = "magenta")

plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.show()
```

label paramter, apply label to the chart

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.step(x, y, marker = "o", label = "python")

plt.legend()
plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.show()
```
