# Texts in Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 5, 2, 6]

plt.plot(x, y)

plt.title("Python Growth", fontsize = 20, color = "lightgreen")
plt.xlabel("Days", fontsize = 15)
plt.ylabel("Growth", fontsize = 15)
plt.text(2, 4, # text(where_on_x_axis, where_on_y_axis, text)
        "JAVA", 
        fontsize = 15,
        style = "normal",
        bbox = {"facecolor": "yellow", "alpha": 0.5, "pad": 15} # "pad" is the space between the text and the box
        )

plt.show()
```
