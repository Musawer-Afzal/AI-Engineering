# Texts in Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 5, 2, 6]

plt.plot(x, y)

plt.title("Python Growth", fontsize = 20, color = "lightgreen")
plt.xlabel("Days", fontsize = 15)
plt.ylabel("Growth", fontsize = 15)
plt.text(2, 4, # text(where_on_x_axis, where_on_y_axis, text)
        "JAVA", 
        fontsize = 15,
        style = "normal",
        bbox = {"facecolor": "yellow", "alpha": 0.5, "pad": 15} # "pad" is the space between the text and the box
        )

plt.show()
```

**plt.annotate**, Write a text inside a plot and control its properties and where it points too

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 5, 2, 6]

plt.plot(x, y)

plt.title("Python Growth", fontsize = 20, color = "lightgreen")
plt.xlabel("Days", fontsize = 15)
plt.ylabel("Growth", fontsize = 15)
plt.text(2, 4, # text(where_on_x_axis, where_on_y_axis, text)
        "JAVA", 
        fontsize = 15,
        style = "normal")

plt.annotate(
    "Python", 
    xy = (3, 2), # This is where the text's arrow will point too
    xytext = (4, 4), # This is where the arrows tails would be
    arrowprops = dict(arrowstyle = "->", color = "red", lw = 2, alpha = 0.5))


plt.show()
```

**plt.legend**

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 5, 2, 6]

plt.plot(x, y)

plt.title("Python Growth", fontsize = 20, color = "lightgreen")
plt.xlabel("Days", fontsize = 15)
plt.ylabel("Growth", fontsize = 15)

plt.legend(["UP"], loc = 9)


plt.show()
```

- facecolor parameter in legend, to change the color of label
- edgecolor parameter, to change the edge color of label

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 5, 2, 6]

plt.plot(x, y)

plt.title("Python Growth", fontsize = 20, color = "lightgreen")
plt.xlabel("Days", fontsize = 15)
plt.ylabel("Growth", fontsize = 15)

plt.legend(["UP"], loc = 9, facecolor = "Red", edgecolor = "magenta")


plt.show()
```
