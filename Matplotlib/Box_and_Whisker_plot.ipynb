# Box Plot

```python
import matplotlib.pyplot as plt
```

```python
x = [10, 20, 30, 40, 50, 60, 70]

plt.boxplot(x)
plt.show()
```

**Whiskey plot is actually a part of Box plot. The upper and lower part of a box plot is called whiskey plot.**

notch parameter, change how the plot shows

```python
x = [10, 20, 30, 40, 50, 60, 70]


plt.boxplot(x, notch = True)


plt.show()
```

```python
x = [10, 20, 30, 40, 50, 60, 70]


plt.boxplot(x, vert = False)


plt.show()
```

widths parameter, change the width of the plot

```python
x = [10, 20, 30, 40, 50, 60, 70]


plt.boxplot(x, widths = 0.8)


plt.show()
```

labels parameter, change the label that appears at the box plot. It is always preferred to use a list to declare the labels as almost all the time we are going to be plotting multiple plots and lists will structure them preperly

```python
x = [10, 20, 30, 40, 50, 60, 70]


plt.boxplot(x, labels = ["python"])


plt.show()
```

patch_artist parameter, fill the color of the box

```python
x = [10, 20, 30, 40, 50, 60, 70]


plt.boxplot(x, patch_artist = True)


plt.show()
```

showmeans parameter, show the mean duh

```python
x = [10, 20, 30, 40, 50, 60, 70]


plt.boxplot(x, showmeans = True)


plt.show()
```

how outliers are shown

```python
x = [10, 20, 30, 40, 50, 60, 70, 120]


plt.boxplot(x)


plt.show()
```

whis parameter, we can use this to join the outlier to the box plot using the line

```python
x = [10, 20, 30, 40, 50, 60, 70, 120]


plt.boxplot(x, whis = 2)


plt.show()
```

sym parameter, change the shape and color of outlier. If you want to hide the outlier then leave the **space** inside quotation marks. If you do not fill the space completely by Box plot then leave the **space** blank inside quotation marks

```python
x = [10, 20, 30, 40, 50, 60, 70, 120]


plt.boxplot(x, sym = " ")


plt.show()
```

boxprops parameter, we can change the color of the box plot. Use dict to declare the changes

```python
x = [10, 20, 30, 40, 50, 60, 70, 120]


plt.boxplot(x, boxprops = dict(color = "r"))


plt.show()
```

capprops parameter, change the properties of caps

```python
x = [10, 20, 30, 40, 50, 60, 70]


plt.boxplot(x, capprops = dict(color = "magenta"))


plt.show()
```

wiskerprops parameter, change the properties of whisker

```python
x = [10, 20, 30, 40, 50, 60, 70]


plt.boxplot(x, whiskerprops = dict(color = "magenta"))


plt.show()
```

fillerprops parameter

```python
x = [10, 20, 30, 40, 50, 60, 70, 120]


plt.boxplot(x, flierprops = dict(markeredgecolor = "magenta"))


plt.show()
```

```python
x = [10, 20, 30, 40, 50, 60, 70]
x1 = [10, 20, 40, 50, 30, 60, 90]

y = [x, x1]

plt.boxplot(y, labels = ["python", "c++"])


plt.show()
```
