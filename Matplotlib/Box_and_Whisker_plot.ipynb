# Box Plot

```python
import matplotlib.pyplot as plt
```

```python
x = [10, 20, 30, 40, 50, 60, 70]

plt.boxplot(x)
plt.show()
```

**Whiskey plot is actually a part of Box plot. The upper and lower part of a box plot is called whiskey plot.**

notch parameter, change how the plot shows

```python
x = [10, 20, 30, 40, 50, 60, 70]


plt.boxplot(x, notch = True)


plt.show()
```
