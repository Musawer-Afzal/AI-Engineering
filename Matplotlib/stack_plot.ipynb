# Stack Plot

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
area1 = [2, 3, 2, 5, 4]
area2 = [5, 7, 3, 9, 4]
area3 = [3, 5, 4, 6, 7]

plt.stackplot(x, area1, area2, area3)


plt.show()
```

For multiple labels it is better to declare the labels beforehand in a list

```python
x = [1, 2, 3, 4, 5]
area1 = [2, 3, 2, 5, 4]
area2 = [5, 7, 3, 9, 4]
area3 = [3, 5, 4, 6, 7]

l = ["area1", "area2", "area3"]
plt.stackplot(x, area1, area2, area3, labels = l)

plt.legend(loc = 2) # loc goes upto 10 with every integer giving a seperate location
plt.show()
```

colors parameter, change the colors

```python
x = [1, 2, 3, 4, 5]
area1 = [2, 3, 2, 5, 4]
area2 = [5, 7, 3, 9, 4]
area3 = [3, 5, 4, 6, 7]

l = ["area1", "area2", "area3"]
c = ["r", "g", "magenta"]
plt.stackplot(x, area1, area2, area3, labels = l, colors = c)

plt.legend(loc = 2)
plt.show()
```
