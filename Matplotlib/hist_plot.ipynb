```python
number = [31, 27, 41, 22, 54, 37, 48, 25, 12, 28, 49, 53, 22, 45, 29, 33, 50, 45, 53, 24, 41, 16, 29, 14, 46, 34, 52, 22, 48, 51, 29, 15, 34, 28, 37, 19, 41, 45, 25, 55, 26, 11, 45, 16, 45, 25, 46, 21, 26, 56]
l = [10, 20, 30, 40, 50, 60]

plt.hist(
    number, 
    color = "blue", 
    edgecolor = "black",
    cumulative = -1,
)

plt.title("Histogram", fontsize = 20)
plt.xlabel("Python", fontsize = 15)
plt.ylabel("Number", fontsize = 15)
plt.show()
```

### Types of histogram

- The histogram we worked on up there is called solid histogram

```python
number = [31, 27, 41, 22, 54, 37, 48, 25, 12, 28, 49, 53, 22, 45, 29, 33, 50, 45, 53, 24, 41, 16, 29, 14, 46, 34, 52, 22, 48, 51, 29, 15, 34, 28, 37, 19, 41, 45, 25, 55, 26, 11, 45, 16, 45, 25, 46, 21, 26, 56]
l = [10, 20, 30, 40, 50, 60]

plt.hist(
    number, 
    color = "blue", 
    edgecolor = "black",
    bins = l,
    histtype = "step" # "step", "stepfilled", "bar", "barfilled", "barstacked"
)

plt.title("Histogram", fontsize = 20)
plt.xlabel("Python", fontsize = 15)
plt.ylabel("Number", fontsize = 15)
plt.show()
```

Orientation parameter, we can show the bars as vertical(default) or horizontal

```python
number = [31, 27, 41, 22, 54, 37, 48, 25, 12, 28, 49, 53, 22, 45, 29, 33, 50, 45, 53, 24, 41, 16, 29, 14, 46, 34, 52, 22, 48, 51, 29, 15, 34, 28, 37, 19, 41, 45, 25, 55, 26, 11, 45, 16, 45, 25, 46, 21, 26, 56]
l = [10, 20, 30, 40, 50, 60]

plt.hist(
    number, 
    color = "blue", 
    edgecolor = "black",
    bins = l,
    # orientation = "horizontal" # By defailt it is vertical
)

plt.title("Histogram", fontsize = 20)
plt.xlabel("Python", fontsize = 15)
plt.ylabel("Number", fontsize = 15)
plt.show()
```

rwidth, change the width of the bars

```python
number = [31, 27, 41, 22, 54, 37, 48, 25, 12, 28, 49, 53, 22, 45, 29, 33, 50, 45, 53, 24, 41, 16, 29, 14, 46, 34, 52, 22, 48, 51, 29, 15, 34, 28, 37, 19, 41, 45, 25, 55, 26, 11, 45, 16, 45, 25, 46, 21, 26, 56]
l = [10, 20, 30, 40, 50, 60]

plt.hist(
    number, 
    color = "blue", 
    edgecolor = "black",
    bins = l,
    rwidth = 0.5
)

plt.title("Histogram", fontsize = 20)
plt.xlabel("Python", fontsize = 15)
plt.ylabel("Number", fontsize = 15)
plt.show()
```

log, change the values in the table to show in the form of log

```python
number = [31, 27, 41, 22, 54, 37, 48, 25, 12, 28, 49, 53, 22, 45, 29, 33, 50, 45, 53, 24, 41, 16, 29, 14, 46, 34, 52, 22, 48, 51, 29, 15, 34, 28, 37, 19, 41, 45, 25, 55, 26, 11, 45, 16, 45, 25, 46, 21, 26, 56]
l = [10, 20, 30, 40, 50, 60]

plt.hist(
    number, 
    color = "blue", 
    edgecolor = "black",
    bins = l,
    log = True
)

plt.title("Histogram", fontsize = 20)
plt.xlabel("Python", fontsize = 15)
plt.ylabel("Number", fontsize = 15)
plt.show()
```

label, give label

```python
number = [31, 27, 41, 22, 54, 37, 48, 25, 12, 28, 49, 53, 22, 45, 29, 33, 50, 45, 53, 24, 41, 16, 29, 14, 46, 34, 52, 22, 48, 51, 29, 15, 34, 28, 37, 19, 41, 45, 25, 55, 26, 11, 45, 16, 45, 25, 46, 21, 26, 56]
l = [10, 20, 30, 40, 50, 60]

plt.hist(
    number, 
    color = "blue", 
    edgecolor = "black",
    bins = l,
    label = "Python"
)

plt.title("Histogram", fontsize = 20)
plt.xlabel("Python", fontsize = 15)
plt.ylabel("Number", fontsize = 15)
plt.legend(loc = "upper left")
plt.show()
```

**axvline**, draw a line in the graph where we want

```python
number = [31, 27, 41, 22, 54, 37, 48, 25, 12, 28, 49, 53, 22, 45, 29, 33, 50, 45, 53, 24, 41, 16, 29, 14, 46, 34, 52, 22, 48, 51, 29, 15, 34, 28, 37, 19, 41, 45, 25, 55, 26, 11, 45, 16, 45, 25, 46, 21, 26, 56]
l = [10, 20, 30, 40, 50, 60]

plt.hist(
    number, 
    edgecolor = "black",
    bins = l,
    label = "Python"
)

plt.title("Histogram", fontsize = 20)
plt.xlabel("Python", fontsize = 15)
plt.ylabel("Number", fontsize = 15)
plt.axvline(x = 45, color = "red", label = "Average")
plt.legend(loc = "upper left")
plt.show()
```
