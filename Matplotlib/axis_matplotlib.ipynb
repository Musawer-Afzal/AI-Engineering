# Axis

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 5, 2, 4]

plt.plot(x, y)

plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 5, 2, 4]

plt.plot(x, y)

plt.xticks(x)
plt.yticks(y)

plt.show()
```

labels parameter, add label to every value

```python
x = [1, 2, 3, 4]
y = [3, 1, 5, 2]

plt.plot(x, y)

plt.xticks(x, labels = ["python", "c", "c++", "jave"])
plt.yticks(y)

plt.show()
```
