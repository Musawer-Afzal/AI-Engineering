# Data Visualization in Python

- Data visualization is a quick, easy way to convey concepts in a universal manner
- It is a graphical representation of information and data
- Visual elements like charts, graphs, and maps

### Plotting libraries:

**Matplotlin:** low level, provides lots of freedom

**Pandas Visualization:** easy to use interface, build on Matplotlib

**Seaborn:** high-level interface, great default styles

**ggplot:** based on R's ggplot2, uses Grammer of Graphics

**Plotly:** can create interactive plots

Types of plot we are going to cover
- Linear Plot
- Scatter Plot
- Bar Plot
- Stem Plot
- Step Plot
- Hist Plot
- Box Plot
- Pie Plot
- Fill_between Plot

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]

c = ["red", "yellow", "green", "blue"]

plt.bar(x, y, color = c)
plt.show()
```

## Bar Plot

```python
x = ["python", "C", "C++", "Java"]
y = [85, 70, 60, 82]
z = [20, 30, 40, 50]
plt.xlabel("Languages", fontsize = 15)
plt.ylabel("Popularity", fontsize = 15)
plt.title("Programming Language Popularity",fontsize = 20)
c = ["red", "yellow", "green", "blue"]
plt.bar(x, 
        y, 
        width = 0.6, 
        color = "red",
        # align = "center", # Change the position of the values under the bar 
        # edgecolor = "black", # Change the color of the edges of the bars
        # linewidth = 2, # Change the width of the edges of the bars
        # linestyle = "--", # What kind of line do you want to use
        # alpha = 0.5 # Change the transparency of the whole chart
        label = "Popularity")
plt.bar(x, 
        z, 
        width = 0.6, 
        color = "yellow", 
        label = "Popularity1")
plt.legend() # Legend is used with label in plt.bar or else lable won't work
plt.show()
```

```python
x = ["python", "C", "C++", "Java"]
y = [85, 70, 60, 82]
z = [20, 30, 40, 50]
p = np.arange(len(x))

width = 0.2
p1 = [j + width for j in p]

plt.xlabel("Languages", fontsize = 15)
plt.ylabel("Popularity", fontsize = 15)
plt.title("Programming Language Popularity", fontsize = 20)

plt.bar(p, 
        y, 
        width, 
        color = "red",
        label = "Popularity")
plt.bar(p1, 
        z, 
        width, 
        color = "yellow", 
        label = "Popularity1")
plt.xticks(p + width / 2, # To show the names of the languages rather than the numbers
           x, # The names we want to give
           rotation = 45) # To rotate the names according to a certain angle
plt.legend(loc = "upper left", # To place the legend on the upper left, bbox_to_anchor has priority
           bbox_to_anchor = (1, 1)) # To place the legend outside the chart
plt.show()
```

Horizontal bar Graph

```python
x = ["python", "C", "C++", "Java"]
y = [85, 70, 60, 82]
z = [20, 30, 40, 50]

plt.xlabel("Languages", fontsize = 15)
plt.ylabel("Popularity", fontsize = 15)
plt.title("Programming Language Popularity",fontsize = 20)

plt.barh(x, 
        y, 
        color = "red",
        label = "Popularity")
plt.barh(x, 
        z, 
        color = "yellow", 
        label = "Popularity1")

plt.legend()
plt.show()
```

```python
x = ["python", "C", "C++", "Java"]
y = [85, 70, 60, 82]
z = [20, 30, 40, 50]
p = np.arange(len(x))

width = 0.2
p1 = [j + width for j in p]

plt.xlabel("Languages", fontsize = 15)
plt.ylabel("Popularity", fontsize = 15)
plt.title("Programming Language Popularity", fontsize = 20)

plt.barh(p, 
        y, 
        width, 
        color = "red",
        label = "Popularity")
plt.barh(p1, 
        z, 
        width, 
        color = "yellow", 
        label = "Popularity1")

# Changes xticks to yticks else it will be all messed up
plt.yticks(p + width / 2, # To show the names of the languages rather than the numbers
           x, # The names we want to give
           rotation = 45) # To rotate the names according to a certain angle
plt.legend(loc = "upper left", # To place the legend on the upper left, bbox_to_anchor has priority
           bbox_to_anchor = (1, 1)) # To place the legend outside the chart
plt.show()
```
